```python
def fetch_kalshi(tickers):
    url = "https://external-api.kalshi.com/trade-api/v2/markets?tickers=" + ",".join(tickers)
    r = requests.get(url, timeout=15)
    out = []
    for m in r.json()["markets"]:
        out.append((
            m["ticker"],
            m["yes_bid_dollars"],
            m["yes_ask_dollars"],
            m["yes_bid_size_fp"],
            m["yes_ask_size_fp"],
        ))
    return out

fetch_kalshi(["KXLALIGAGAME-26SEP15ELCRMA-RMA", "KXLALIGAGAME-26SEP15ELCRMA-TIE"])
```

```python
def fetch_poly(token_ids):
    out = []
    for tid in token_ids:
        try:
            bk = requests.get(f"https://clob.polymarket.com/book?token_id={tid}", timeout=15).json()
        except Exception as e:
            print("poly fetch failed:", e)
            continue

        bids = bk.get("bids", [])
        asks = bk.get("asks", [])
        bb = max(bids, key=lambda x: float(x["price"])) if bids else None
        ba = min(asks, key=lambda x: float(x["price"])) if asks else None

        out.append((
            tid,
            bb["price"] if bb else None,
            ba["price"] if ba else None,
            bb["size"] if bb else None,
            ba["size"] if ba else None,
        ))
    return out
```
